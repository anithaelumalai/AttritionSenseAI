# AttritionSense AI — Random Forest Model Training & Evaluation

This notebook documents the end-to-end Machine Learning pipeline for **AttritionSense AI**:
1. Dataset loading & exploratory inspection (1,470 employees, 35 features).
2. Feature engineering, preprocessing with `ColumnTransformer` (numerical scaling + categorical one-hot encoding).
3. Training a `RandomForestClassifier` with balanced class weights.
4. Comprehensive model evaluation (Accuracy, Precision, Recall, F1-Score, ROC-AUC).
5. Feature importance analysis & artifact serialization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from utils.ml_pipeline import train_and_evaluate_model, load_trained_model

# Load dataset
df = pd.read_csv('../data/employees.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Target distribution
print(df['Attrition'].value_counts(normalize=True))
df['Attrition'].value_counts().plot(kind='bar', color=['#2a9d8f', '#e63946'])
plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Count')
plt.show()
```

```python
# Train Random Forest pipeline and evaluate
pipeline, metadata = train_and_evaluate_model(df)
for metric, val in metadata['metrics'].items():
    if metric != 'confusion_matrix':
        print(f"{metric.upper()}: {val}")
```

```python
# Top 10 feature importances
feats = pd.DataFrame(metadata['feature_ranking'][:10])
plt.figure(figsize=(10, 6))
sns.barplot(data=feats, x='importance', y='feature', palette='viridis')
plt.title('Top 10 Feature Importances in Attrition Prediction')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```
